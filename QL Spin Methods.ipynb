# Quasi-Local Spin Measure Methods
We'll be attempting to use 3 main methods to calculate spin magnitude and direction. This method leaves an up/down ambiguity, which we'll resolve using a very rough surface integral (if no other measures are running) or simply picking the direction closest to that of the other measures. The methods are listed below in order of what should be the simplest/fastest to most complex.

Below is the formula for dimensionful and dimensionless spin *magnitude*. This will be used for methods 2 and 3. The trickier part is generally finding the spin direction.
$$
J = \frac{1}{8\pi} \oint_H \omega_B \phi^B dA \hspace{6mm} \mathrm{dimensionful} \\
\chi = \frac{J}{M^2} \hspace{6mm} \mathrm{dimensionless}
$$

Another generally important formula:
$$
\omega_A = s^j (K_{jb} - \gamma_{jb} K) h^b_A
$$
Where $K$ is extrinsic curvature (should be precalculated), $\gamma$ is the spatial 3-metric, and $h$ is the projector from the 3D spatial slice to the horizon 2-surface (indexed by capital Latin letters)

### Method 1: 2-D Minimization of Circumference Ratios
The ratio $\frac{C_{polar}}{C_{equatorial}}$ will be minimized when the spin vector is going through the poles. We'll evaluate this ratio using the proper circumference calculation from `diagnostics_proper_circumferences.py` at multiple points on the apparent horizon surface, minimizing using the Nelder-Mead method (downhill simplex).

### Method 2: Spin Function ($\Omega$) Surface Integral
$$
\Omega = \epsilon^{AB} \nabla_A \omega_B \\
S^i = \frac{S}{N} \oint_{\mathcal{H}}dA \Omega (x^i - x^i_0-x^i_R) \hspace{6mm} \mathrm{(spin \ vector \ components)}\\
\delta_{ij}S^i S^j = S \hspace{6mm} \mathrm{(N \ makes \ this \ true)}\\
x^i_R = \frac{1}{8\pi} \oint_{\mathcal{H}}dA (x^i - x^i_0)
$$

$R$ is the intrinsic Ricci scalar of the surface, $x^i_0$ are the centroid coordinates.

This can produce dimensionless or dimensionful spin, depending on the magnitude $S$ given

### Method 3: Killing transport
A true Killing vector satisfies $\mathcal{L}_{\xi} g_{\mu\nu} = \nabla_\mu \xi_\nu + \nabla_\nu \xi_\mu = 0$. Since our surfaces are deformed by the pull of the other black hole, there won't be any of these on our actual AH surfaces. Instead, we find an approximate Killing vector (AKV) $\phi$ by minimizing
$$
I(\phi) = \oint_\mathcal{H} |\mathcal{L}_\phi q_{AB}|^2 dA = \oint_\mathcal{H} |\nabla_A \phi_B + \nabla_B \phi_A|^2 dA
$$
My understanding is that ET does this by solving a PDE, but it might be doable with a grid or Nelder-Mead method on the horizon surface.
