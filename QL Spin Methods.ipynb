# Quasi-Local Spin Measure Methods
We'll be attempting to use 3 main methods to calculate spin magnitude and direction. This method leaves an up/down ambiguity, which we'll resolve using a very rough surface integral (if no other measures are running) or simply picking the direction closest to that of the other measures. The methods are listed below in order of what should be the simplest/fastest to most complex.

### Method 1: 2-D Minimization of Circumference Ratios
The ratio $\frac{C_{polar}}{C_{equatorial}}$ will be minimized when the spin vector is going through the poles. We'll evaluate this ratio using the proper circumference calculation from `diagnostics_proper_circumferences.py` at multiple points on the apparent horizon surface, minimizing using the Nelder-Mead method (downhill simplex).
